```python
import requests
from bs4 import BeautifulSoup


def decode_message(url):
    """Print the secret grid of characters published in a Google Doc.

    The doc holds a table where each row is (x, character, y). We read those
    triples, drop them onto a grid, and print it. Row 0 is the top of the
    page, so the largest y goes on the first line.
    """
    html = requests.get(url).text
    soup = BeautifulSoup(html, "html.parser")

    # Skip the header row, then grab the three cells from each remaining row.
    points = []
    for row in soup.find_all("tr")[1:]:
        x, char, y = (cell.get_text(strip=True) for cell in row.find_all("td"))
        points.append((int(x), char, int(y)))

    if not points:
        print("No data found.")
        return

    width = max(x for x, _, _ in points)
    height = max(y for _, _, y in points)

    # Build empty grid, fill in the characters, flip vertically for printing.
    grid = [[" "] * (width + 1) for _ in range(height + 1)]
    for x, char, y in points:
        grid[y][x] = char

    for line in reversed(grid):
        print("".join(line))


if __name__ == "__main__":
    decode_message(
       "https://docs.google.com/document/d/e/2PACX-1vSvM5gDlNvt7npYHhp_XfsJvuntUhq184By5xO_pA4b_gCWeXb6dM6ZxwN8rE6S4ghUsCj2VKR21oEP/pub"
    )
```
